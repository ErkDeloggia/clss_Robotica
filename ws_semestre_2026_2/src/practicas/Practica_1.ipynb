## Previo de la Práctica 1


 ## URDF - Unified Robot Description Format.

 La complejidad de los sistemas robóticos está aumentando, lo que hace que sea cada vez más eficaz simularlos antes de su despliegue. Para ello, se requiere un modelo de la cinemática o dinámica del robot, y el formato más utilizado es el **Unified Robot Description Format (URDF)**.


El **Unified Robot Description Format (URDF)** es un formato de archivo basado en el estándar XML que se utiliza principalmente en el ecosistema **ROS (Robot Operating System)** para describir de manera completa y jerárquica la estructura física de un robot.

Su concepto central se basa en describir al robot como un árbol de componentes conectados entre sí, permitiendo que el software de control y simulación entienda cómo es y cómo se mueve la máquina.

#### ¿Para qué sirve?

Un archivo URDF funciona como un “plano digital” o “esqueleto” que permite a simuladores, herramientas de visualización y algoritmos de control comprender cómo está construido el hardware real. Imagina que es el instructivo de Lego de tu robot: define dónde tiene ruedas, cómo se mueven sus brazos, cuánto pesan sus piezas y qué forma tiene para que los simuladores puedan interactuar con él sin romper nada en el mundo real.

Define cuatro pilares fundamentales: 

***1. Cinemática y geometría:*** Especifica las dimensiones de cada sección y cómo se conectan.

***2. Propiedades dinámicas e inerciales:*** Detalla la masa, el centro de gravedad y el tensor de inercia para simular físicas precisas.

***3. Aspecto visual:*** Modela la apariencia en entornos 3D mediante formas primitivas (cajas, cilindros) o mallas complejas de software CAD .

***4. Modelos de colisión:*** Crea geometrías simplificadas que envuelven al robot para calcular impactos de manera eficiente sin sobrecargar el procesador. 

### Estructura central: Links y Joints

La topología de un URDF se organiza como un *árbol cinemático*. Todo el robot se desglosa en dos elementos esenciales:

### 1. Eslabones (`<link>`)

Representan las partes rígidas o cuerpos del robot (por ejemplo: la base, un brazo articulado, una cámara). Cada eslabón puede incluir geometría visual (cómo se ve el robot en simulación), geometría de colisión (formas simplificadas para cálculos de física) y propiedades inerciales (masa y tensores de inercia para dinámica). 

### Elementos primitivos en URDF

Los elementos primitivos en URDF describen las **formas geométricas básicas** de un robot. Definen la forma geométrica de los eslabones (links). Se definen dentro de la etiqueta `<geometry>`. Se incluyen dentro de las etiquetas `<visual>` (cómo se ve el robot) y `<collision>` (los límites físicos para calcular choques).

En URDF, un eslabón es una pieza rígida del robot. Como el sistema no sabe qué forma tiene esa pieza en la vida real, usas las primitivas (`<box>`, `<cylinder>`, `<sphere>`) para darle esa forma en el mundo virtual.

Tienen dos funciones principales:

**•	Forma Visual (\<visual>):** Es la silueta que tú ves en la pantalla al abrir el simulador (como Gazebo o RViz).

**•	Forma de Colisión (\<collision>):** Es el volumen "invisible" que usa el simulador para saber si el eslabón choca contra una pared u otra parte del robot.

####Detalle técnico y la estructura XML exacta de cada uno:

#### •	1. \<box> (Caja)

Define un paralelepípedo recto (un cubo o caja de seis caras planas). Es la forma más utilizada en URDF para representar cuerpos principales, chasis de robots móviles o computadoras de a bordo.

**Estructura y Parámetros**

La etiqueta utiliza un único atributo llamado size, el cual requiere tres valores numéricos obligatorios separados por espacios. Estos valores se miden en metros:


  ![box](imagenes_p1/box.png)

#### Figura 1. Configuracion \<box> 

**X (Largo):** Dimensión en el eje rojo de ROS.

**Y (Ancho):** Dimensión en el eje verde de ROS.

**Z (Alto):** Dimensión en el eje azul de ROS.

**Origen de Coordenadas**

Por defecto, el origen (0 0 0) del sistema de coordenadas se ubica exactamente en el **centro geométrico** de la caja.

Si defines una caja de size="1 1 1", la mitad de la caja estará en la zona positiva de los ejes y la otra mitad en la zona negativa (se extiende 0.5 metros en cada dirección desde el centro).

**¿Cuándo debes usarlo?**

•	Para crear el chasis de un robot móvil (AGV o AMR).

•	Para modelar los eslabones cuadrados de un brazo robótico.

•	Para representar sensores como cámaras o LIDARs de forma simplificada.


#### 2. \<cylinder> (Cilíndro) 

Representa un cilindro circular. Sus dimensiones se configuran mediante los atributos radius y length (ambos en metros). Al igual que la caja, su origen está en el centro geométrico del cilindro, y su eje de simetría se alinea por defecto con el eje (Z)
El elemento primitivo \<cylinder> en URDF se utiliza para representar cuerpos cilíndricos rectos (como tubos, ejes, cilindros neumático-hidráulicos o ruedas). Al igual que la caja, es una pieza fundamental para construir la cinemática de un robot en ROS 2.

**1. Estructura XML y Reglas de Sintaxis**
La estructura de un cilindro se define mediante dos atributos geométricos obligatorios dentro de la etiqueta \<geometry>:

  ![cylinder](imagenes_p1/cylinder.png)

#### Figura 2. Configuracion \<cylinder> 

**Atributos obligatorios:**

•	**radius (Radio):** La distancia (en metros) desde el centro del cilindro hasta su borde circular exterior.

•	**length (Longitud/Altura):** La dimensión total (en metros) a lo largo del eje del cilindro.

**Restricciones críticas:**

•	Ambos valores deben ser estrictamente números reales flotantes positivos mayores a cero.

•	No se permiten comas como separadores de atributos.

•	Su centro geométrico (0 0 0) se sitúa exactamente en la mitad de su longitud y en el centro de su sección transversal circular.

**2. Características Geométricas y Cinemáticas Avanzadas**
   
**Orientación por Defecto en ROS 2 (La Regla del Eje Z)**

Por diseño por defecto en URDF, todos los cilindros se crean alineados verticalmente a lo largo del eje Z local del link. La longitud se extiende hacia arriba (+Z) y hacia abajo (-Z) desde su origen geométrico.

Si necesitas usar un cilindro en posición horizontal (por ejemplo, para representar el eje de un motor o una rueda), tienes que rotarlo explícitamente usando el atributo rpy (Roll, Pitch, Yaw) en la etiqueta \<origin>.

**•	Para alinearlo con el eje X:** Rótalo (90°) en Pitch (($\pi$/2) radianes)

  ![cylinder1](imagenes_p1/cylinder1.png)

#### Figura 3. Configuracion alineado con el eje X

**Para alinearlo con el eje Y (Típico para ruedas):**  Rótalo (90°) en Roll.

  ![cylinder1](imagenes_p1/cylinder1.png)

#### Figura 4. Configuracion alineado con el eje Y

**3. Aplicaciones Reales en la Industria de la Robótica**

**A. Ruedas de Robots Móviles (AGVs / AMRs / Drones)**

Es la aplicación más común. Las ruedas motrices y las ruedas locas (caster wheels) de chasis industriales se modelan con cilindros de gran radio y corta longitud. Al rotarlos sobre el eje Y o X, interactúan de forma nativa con las juntas de tipo continuous de ROS 2.

**B. Eslabones de Brazos Manipuladores**

La gran mayoría de los eslabones (links) que conectan las articulaciones de un brazo robótico (como un robot antropomórfico o un brazo SCARA) son cilindros o tubos. Ofrecen un balance simétrico idóneo para rotaciones repetitivas de alta velocidad en los motores.

**C. Columnas de Elevación y Actuadores Lineales**
En sistemas de paletizado automático o robots móviles que ajustan su altura para levantar estanterías, las guías telescópicas y los pistones neumáticos se representan mediante cilindros delgados con un length considerable. Estos eslabones se conectan mediante juntas tipo prismatic (prismáticas o deslizantes).

**D. Modelado Simplificado de Sensores LiDAR**

Los sensores de escaneo láser en 2D y 3D (como los sensores Velodyne, Ouster o Hokuyo) tienen carcasas cilíndricas que rotan internamente. Representar su volumen exterior con un \<cylinder> en la geometría de colisión protege al sensor de impactos contra obstáculos en el entorno virtual.

#### 3. Sphere (Esfera)

Representa una esfera perfecta. Su tamaño se define mediante el atributo radius (en metros). El origen se sitúa exactamente en su centro.

El elemento primitivo \<sphere> es la forma más simple y simétrica en URDF. Se utiliza para representar geometrías perfectamente esféricas, lo que ofrece ventajas matemáticas únicas tanto en la detección de colisiones como en el cálculo de dinámicas para ROS 2.

**1. Estructura XML y Reglas de Sintaxis**

La estructura de una esfera en URDF es sumamente compacta. Al ser un objeto con simetría radial perfecta, solo requiere un único atributo geométrico dentro de la etiqueta \<geometry>:


  ![sphere](imagenes_p1/sphere.png)

#### Figura 5. Configuracion \<sphere> 

**Atributo único obligatorio:**

**•	radius (Radio):** La distancia (en metros) desde el centro de la esfera hasta cualquier punto de su superficie externa.

**Restricciones críticas:**

•	El valor de radius debe ser un número real flotante estrictamente positivo mayor a cero.

•	Su origen nativo (0 0 0) se ubica exactamente en el centro geométrico de la esfera. Por ello, no importa cuánto cambies la rotación (rpy) en la etiqueta \<origin>, visualmente la esfera no experimentará ningún cambio de orientación.

**2. Características Geométricas y Cinemáticas Avanzadas**

**Simetría de Rotación e Invariabilidad**

A diferencia del cilindro (que requiere rotarse para acostarlo) o de la caja (que cambia de posición según sus caras), la esfera es invariante ante las rotaciones en sus ejes locales. Esto simplifica drásticamente las transformaciones matemáticas en los nodos de ROS 2 (tf2).

**3. Aplicaciones Reales en la Industria de la Robótica**

**A. Ruedas Locas (Caster Wheels)**

En la base de los robots móviles de almacén (AGVs), además de las ruedas motrices cilíndricas, se instalan ruedas locas de apoyo para mantener el equilibrio. En la vida real suelen ser ruedas con horquillas giratorias, pero simular ese mecanismo añade fricciones e inestabilidades innecesarias.

•	Solución industrial: Se modela el apoyo como una esfera fija de baja fricción en la base del URDF. Esto permite que el robot se deslice suavemente en cualquier dirección (X o Y) sin alterar la odometría.

**B. Simplificación Extrema de Colisiones (Bounding Spheres)**

Cuando un brazo robótico industrial o las falanges de una mano robótica antropomórfica se mueven en espacios reducidos, calcular si van a chocar contra el entorno requiere mucha potencia gráfica.

**•	Técnica avanzada:** Se envuelven las articulaciones complejas o las puntas de las pinzas (grippers) dentro de una \<sphere> invisible en el bloque \<collision>. Detectar si una esfera toca otro objeto es la operación matemática de colisión más rápida que existe (solo calcula si la distancia entre dos puntos es menor que la suma de sus radios).

**C. Herramientas de Extremo (End Effectors) y Sondas**

Muchos robots de medición por coordenadas (CMM) o brazos articulados de inspección utilizan palpadores con puntas de rubí esféricas. Representar la punta de la herramienta con una \<sphere> permite lecturas de contacto exactas en simulación desde cualquier ángulo de aproximación.

**D. Modelado de Articulaciones Esféricas (Ball and Socket)**

Aunque URDF no admite de forma nativa juntas de tipo rótula o esféricas directas (se deben construir combinando tres juntas rotacionales continuas perpendiculares), los eslabones que forman el núcleo físico de estas articulaciones se diseñan usando esferas para garantizar que las carcasas no colisionen entre sí durante movimientos angulares extremos.

### 2. Articulaciones (`<joint>`)

Las articulaciones determinan las conexiones entre eslabones y especifican cómo pueden moverse relativamente entre sí. Asimismo, definen la relación matemática y de movimiento entre un eslabón **padre** (*parent*) y un eslabón **hijo** (*child*). Dados dos eslabones unidos por una articulación, el más cercano a la raiz de la estructura (es decir a la base o al chásis del robot) recibe el nombre de **padre** y el otro se denomina **hijo**.

El URDF admite varios tipos de articulaciones: 

#### 1. Fixed

Bloquea el movimiento entre ambos eslabones (sin grados de libertad).  Básicamente, suelda o pega rígidamente ambos eslabones para que actúen como un solo cuerpo sólido.

**Características principales**

**•	Cero grados de libertad:** No permite rotación ni traslación (0 DOF).

**•	Conexión rígida:** Fusiona dos geometrías en un solo cuerpo rígido.

**•	Sin parámetros mecánicos:** No requiere límites de esfuerzo, velocidad ni fricción.

**•	Transformación constante:** La distancia y orientación entre ambos eslabones nunca cambia.

  ![fixed](imagenes_p1/fixed.png)

#### Figura 6. Configuracion fixed 

**Para qué sirve y aplicaciones**

Aunque no permite movimiento, es fundamental en robótica para estructurar el árbol cinemático del robot por las siguientes razones:

**•	Añadir sensores estáticos:** Para fijar cámaras (RGB-D), sensores LiDAR o de ultrasonido al chasis del robot en una posición exacta.

**•	Dividir geometrías complejas:** Si la base de tu robot tiene una forma muy compleja, puedes modelarla en piezas separadas en URDF y unirlas rígidamente con juntas fixed.

**•	Puntos de referencia ficticios:** Para crear un eslabón invisible sin peso en el centro del robot (como base_link o base_footprint) que sirve como origen de coordenadas globales para la navegación.

**Comportamiento en simulación**

 **+ En RViz / ROS:**

**-Función cinemática:** RViz solo visualiza la estructura del robot mediante el paquete robot_state_publisher.

**-Visualización:** Lee el archivo URDF y mantiene los eslabones perfectamente unidos.

**-Árbol de transformaciones:** Publica una transformación estática (tf) constante entre los dos eslabones.

**-Consumo de recursos:** Es sumamente ligero porque no realiza cálculos matemáticos de movimiento.

**+ En simuladores dinámicos (Gazebo / KDL):** 

**-Fusión de enlaces:** Los motores de física (ODE, Bullet) fusionan los dos eslabones en una sola masa rígida.

**-Propiedades físicas:** Suma automáticamente las masas y calcula el nuevo centro de inercia combinado.

**-Reducción de ecuaciones:** Elimina las restricciones de la articulación para ahorrar poder de cómputo.

**-Errores comunes:** Si las inercias de los eslabones fijados están mal calculadas, el robot puede vibrar o salir disparado (efecto glitch).

#### 2. Revolute: 

Rotación sobre un eje con límites angulares específicos. (un ángulo mínimo y un ángulo máximo).
Es el equivalente a un codo humano, una rodilla o el dedo de una pinza robótica.

**Características principales**

**•	Un solo grado de libertad (1 DoF):** Solo rota en un eje (x), (y) o (z).

**•	Rotación limitada:** No puede girar de forma infinita (por ejemplo, solo se mueve de (-90) a (+90).

**•	Parámetros obligatorios:** Requiere especificar los límites físicos, el esfuerzo máximo (torque) y la velocidad máxima.

  ![revolute](imagenes_p1/revolute.png)

#### Figura 7. Configuracion revolute 

La articulación revolute se utiliza principalmente para modelar cualquier unión de un robot que imite el movimiento de un mecanismo de bisagra o una articulación biológica con un rango de movimiento restringido.

**Sus aplicaciones más comunes en robótica incluyen:**

**1. Brazos y Manipuladores Robóticos**

Es el tipo de junta más utilizado en los brazos industriales (como los brazos de 6 ejes o los robots colaborativos):

**•	Hombros, codos y muñecas:** Permiten que el brazo se extienda y gire para alcanzar objetos, pero con límites físicos para evitar que los cables internos se rompan o que el robot choque contra su propia estructura.

**•	Pinzas / Efectores finales (Grippers):** Para controlar la apertura y cierre de los dedos mecánicos que sujetan los objetos.

**2. Robótica Humanoide y Biomimética**

Se usa para replicar casi todas las articulaciones del cuerpo humano o de animales:

**•	Extremidades:** Rodillas, tobillos y codos.

**•	Cuello y cabeza:** Para permitir que el robot mire hacia los lados o arriba y abajo, limitando el giro para que no dé una vuelta completa de (360°).

**3. Sistemas de Dirección de Vehículos**

**•	Dirección Ackermann:** En robots móviles con ruedas (tipo automóvil), se usa para hacer girar las ruedas delanteras hacia la izquierda o derecha para cambiar el rumbo, controlando con precisión el ángulo máximo de giro.

**4. Mecanismos de Movimiento Auxiliar**

**•	Sistemas Pan-and-Tilt:** Las bases móviles que orientan cámaras o sensores LiDAR para escanear el entorno.

**•	Tapas, compuertas o alerones:** Cualquier parte del robot que deba abrirse o cerrarse hasta un tope físico.

**Comportamiento en simulación**

**1. Comportamiento en RViz (ROS)**

RViz no tiene motor de física; solo se encarga de la representación gráfica 3D y las transformaciones matemáticas en el espacio.

**•	Cinemática pura:** RViz lee la posición de la articulación a través del nodo robot_state_publisher usando mensajes de tipo sensor_msgs/JointState. Simplemente "dibuja" el eslabón en el ángulo exacto que recibe.

**•	Sin gravedad ni colisiones:** Si no envías datos de posición, la articulación se quedará flotando estáticamente en el ángulo 0.0, ignorando el peso del eslabón. Tampoco detectará si una pieza atraviesa a otra.

**•	Pruebas con joint_state_publisher_gui:** Para probarla visualmente sin programar, puedes lanzar este nodo auxiliar. Te mostrará una barra deslizante (slider) limitada estrictamente entre los valores lower y upper que definiste en el URDF para que muevas la articulación manualmente.

**2. Comportamiento en Gazebo / Simuladores Físicos**

Gazebo utiliza un motor de física dinámico (como ODE o Bullet). Aquí el robot interactúa con el entorno, la gravedad y las fuerzas.

**•	Dinámica real:** La articulación tiene masa e inercia (calculadas desde la etiqueta \<inertial> del eslabón). Si el motor no aplica fuerza, la articulación caerá libremente por efecto de la gravedad.

**•	Paredes físicas (Límites):** Al llegar a los límites lower o upper, el motor de física simula un choque contra un tope sólido. Si la fuerza del impacto es muy alta, el robot puede rebotar o desestabilizarse.

**•	Saturación por esfuerzo (effort):** Si intentas levantar un objeto muy pesado y el torque necesario supera el límite de effort que pusiste en el URDF, el simulador no moverá la articulación, imitando a un motor real que se ha quedado sin potencia.

**•	Fricción del mundo real:** Mediante la etiqueta \<dynamics>, Gazebo simula la fricción del eje (friction) y el amortiguamiento del motor (damping), haciendo que el movimiento se detenga gradualmente por resistencia física en lugar de girar infinitamente por inercia.


#### 3. Continuous 

Es una junta de rotación que permite el movimiento de giro continuo e infinito alrededor de un solo eje, sin ningún tipo de restricción angular. (ideal para ruedas de robots móviles).

Es el equivalente exacto a un eje rotatorio libre, como el de un motor de corriente directa o un rodamiento.

**Características principales**

**•	Sin límites angulares:** No tiene topes físicos; puede girar desde (-infty) hasta \(+infty ) radianes.

**•	Un solo grado de libertad (1 DoF):** Su movimiento se restringe a la rotación sobre el eje específico definido (x), (y) o (z).

**•	Sintaxis simplificada:** A diferencia de la articulación revolute, en el código XML de la junta continuous no se deben incluir los atributos lower y upper dentro de la etiqueta \<limit>.

**•	Parámetros dinámicos opcionales:** Puede incluir límites de velocidad (velocity) y torque máximo (effort) para restringir su potencia.

  ![continuous](imagenes_p1/continuous.png)

#### Figura 8. Configuracion continuous 

**¿Para qué sirve? (Aplicaciones comunes)**

Se utiliza para modelar componentes del robot que requieren un giro constante y libre en una sola dirección o en reversa:

**1. Ruedas de robots móviles:** Es su aplicación más común (robots de tracción diferencial, carros autónomos, AGVs). Permite que las ruedas giren indefinidamente para que el robot avance.

**2. Aspas y hélices:** Para modelar los rotores de drones (cuatricópteros) o ventiladores de refrigeración.

**3. Sistemas de escaneo rotativos:** Soporte para sensores LiDAR 2D que giran continuamente a (360°) para mapear el entorno.

**4. Herramientas giratorias:** Brocas de robots de manufactura o husillos de fresado.

**Comportamiento en simulación**

El comportamiento de la junta continuous varía según la herramienta debido a la ausencia de límites angulares:

**En RViz / ROS (Visualizador)**

**•	Giro sin restricciones:** Si utilizas la herramienta joint_state_publisher_gui, el control deslizante (slider) de esta articulación no tiene extremos fijos; al llegar al extremo de la barra, vuelve a empezar en un ciclo infinito de (0) a (2$\pi$) radianes (360°).

**•	Actualización de posición:** RViz simplemente rota la malla 3D del eslabón hijo según la lectura matemática de la odometría o del estado de las articulaciones, sin importar cuántas vueltas haya dado.

**En Gazebo (Simulador Físico)**

**•	Control por velocidad:** Mientras que las juntas revolute se controlan habitualmente por posición, las juntas continuous se controlan casi siempre mediante comandos de velocidad (radianes por segundo).

**•	Inercia de giro:** Al apagar el motor virtual, la rueda o hélice no se detiene en seco. Seguirá girando por un momento debido a la inercia calculada en la etiqueta \<inertial> del eslabón hijo.

**•	Fricción continua:** Es crucial configurar la etiqueta \<dynamics damping="..." friction="..."/>. Si no se añade fricción en el eje, una rueda en el aire a la que se le aplica un torque constante acelerará de forma irreal hasta el infinito.

**•	Interacción con el suelo:** En el caso de las ruedas, Gazebo calcula constantemente las fuerzas de fricción entre el neumático (eslabón hijo) y la superficie del suelo para generar el desplazamiento del robot por tracción.




#### 4. Prismatic 

Es una junta que permite el movimiento de traslación o deslizamiento a lo largo de un solo eje.

Es el equivalente mecánico a un pistón, un riel telescópico o un actuador lineal, y cuenta con límites físicos obligatorios de principio y fin de carrera.

**Características principales**

**•	Un grado de libertad lineal (1 DoF):** El eslabón hijo solo puede desplazarse en línea recta a lo largo del eje definido (x), (y) o (z).

**•	Límites métricos obligatorios:** Requiere especificar obligatoriamente los límites de carrera mínimos y máximos mediante la etiqueta \<limit>. A diferencia de las juntas rotativas, aquí las unidades se miden estrictamente en metros.

**•	Parámetros de fuerza:** Exige definir el esfuerzo máximo (effort), que en este caso se traduce como fuerza (Newtons) en lugar de torque, y la velocidad máxima (velocity en metros por segundo).

  ![prismatic](imagenes_p1/prismatic.png)

#### Figura 9. Configuracion prismatic

**¿Para qué sirve? (Aplicaciones comunes)**

Se utiliza en cualquier mecanismo donde el robot deba extender, contraer, subir o bajar una de sus partes de manera lineal:

**1. Ejes Z en robots industriales o impresoras 3D:** Para elevar o descender el cabezal de impresión, la herramienta de corte o el brazo robótico (como en los robots tipo Cartesiano o SCARA).

**2. Sistemas de elevación (Montacargas / AGVs):** En robots móviles de almacén que necesitan elevar plataformas o palés para transportarlos.

**3. Brazos telescópicos:** Secciones de un manipulador que se extienden hacia adelante para alcanzar objetos lejanos.

**4. Pistones y actuadores lineales:** Mecanismos neumáticos o hidráulicos que empujan o tiran de una pieza de la estructura.

**Comportamiento en simulación**

Al igual que las juntas anteriores, su comportamiento se divide entre el entorno puramente geométrico y el entorno físico:

**En RViz / ROS (Visualizador)**

**•	Desplazamiento visual acotado:** Al utilizar herramientas como joint_state_publisher_gui, el control deslizante (slider) te permitirá desplazar el eslabón linealmente. La barra tendrá como límites exactos los valores en metros configurados en lower y upper.

**•	Cinemática de traslación:** RViz actualiza la matriz de transformación (TF) aplicando una traslación pura sobre el eje especificado, desplazando la malla 3D de forma rígida sin aplicar rotaciones.

**En Gazebo (Simulador Físico)**

**•	Impacto en los topes mecánicos:** Cuando el actuador lineal llega al límite de su carrera (lower o upper), el motor de física de Gazebo detiene el movimiento simulando un choque seco contra un tope físico. Si lleva mucha velocidad, el impacto puede sacudir o desestabilizar la estructura del robot.

**•	Control por Fuerza / Posición:** Se suele controlar mediante un lazo PID virtual que calcula la fuerza necesaria (en Newtons) para mover la masa del eslabón hijo a la distancia deseada.

**•	Efecto de la gravedad:** Si el eje prismático es vertical (como un elevador), el motor de física aplicará la gravedad sobre la masa del eslabón. Si el límite de fuerza (effort) configurado en el URDF es muy bajo, el actuador no tendrá la fuerza suficiente para vencer el peso del propio eslabón y este se caerá.

**•	Amortiguación lineal:** Es vital configurar parámetros de fricción y amortiguamiento en la etiqueta \<dynamics>. Esto evita que el eslabón "tiemble" o vibre de forma irreal dentro del riel virtual al interactuar con fuerzas externas.


#### 5. Planar

Es una junta en URDF que permite al eslabón hijo desplazarse y rotar de forma libre, pero estrictamente restringido a la superficie de un plano bidimensional. permite el movimiento en un plano perpendicular al eje.

Es el equivalente a deslizar una pieza de hockey sobre una mesa de aire o mover un imán sobre una pizarra.

**Características principales**

**Tres grados de libertad (3 DoF):** Permite dos movimientos de traslación (generalmente en los ejes (X) e (Y) y un movimiento de rotación (giro sobre el eje (Z), perpendicular al plano).

**Orientación por eje:** Por defecto, el movimiento ocurre en el plano perpendicular al eje especificado en la etiqueta \<axis>. Si el eje es 0 0 1 (Z), el robot se moverá sobre el plano horizontal (X-Y).

**Límites opcionales:** Permite configurar límites de carrera para las traslaciones mediante la etiqueta \<limit>, aunque en muchas aplicaciones de base móvil se deja libre.


  ![planar](imagenes_p1/planar.png)

#### Figura 10. Configuracion planar 


**¿Para qué sirve? (Aplicaciones comunes)**

Se utiliza principalmente para simplificar el modelado de bases robóticas terrestres o mecanismos que operan sobre superficies totalmente planas:

**1. Bases de Robots Móviles (AGVs / AMRs):** Modela el movimiento de plataformas móviles (como un robot aspirador o un robot de almacén) sobre el suelo de manera directa, omitiendo la complejidad física de las ruedas individuales.

**2. Mesas de coordenadas X-Y:** Plotters, cortadoras láser o mesas de posicionamiento médico donde un cabezal se desplaza horizontalmente en dos ejes y puede o no rotar.

**3. Brazos robóticos Scara o de configuración plana:** Para simplificar el análisis cinemático de los primeros eslabones que solo operan de forma paralela al suelo.

**4. Simulación simplificada de juegos:** Juegos de mesa físicos como el hockey de aire o el billar robótico.

**Comportamiento en simulación**

Su comportamiento equilibra la libertad espacial con restricciones rígidas impuestas por el plano elegido:

**En RViz / ROS (Visualizador)**

**•	Movimiento en 2D:** RViz traslada gráficamente el modelo en dos direcciones y lo hace rotar sobre el plano definido, basándose en las transformadas (TF) de odometría que recibe.

**•	Restricción visual absoluta:** El eslabón hijo jamás se separará del plano en el visualizador, manteniendo una altura constante de forma matemática sin importar el resto del entorno.

**•	Sin sliders por defecto:** Al igual que la junta flotante, las herramientas básicas de interfaz gráfica no generan controles deslizantes automáticos de forma nativa debido a su naturaleza multidimensional de 3 DoF.


**En Gazebo (Simulador Físico)**

**•	Física restringida:** El motor de física permite que el eslabón interactúe con el entorno (choques, empujones, fricción), pero anula cualquier fuerza o componente de gravedad que intente sacarlo del plano. Si un objeto cae sobre él, no se aplastará ni se inclinará hacia abajo.

**•	Control por Velocidad Planar:** En lugar de controlar actuadores individuales, se suele utilizar un plugin de ROS en Gazebo (como un controlador de tracción planar u omnidireccional) para enviar comandos de velocidad cinemática en (X), (Y) y velocidad angular en (Z) (geometry_msgs/Twist).

**•	Ideal para pruebas de algoritmos:** Al eliminar los rebotes milimétricos provocados por la suspensión o el contacto de las ruedas reales con el piso, esta articulación es perfecta para probar algoritmos puros de navegación (SLAM) en un entorno controlado y computacionalmente ligero.




#### 6. Floating

Es una junta especial en URDF que permite un movimiento completamente libre en el espacio tridimensional entre el eslabón padre y el hijo.
A diferencia de las juntas anteriores, esta no restringe ningún movimiento, permitiendo un desplazamiento total en cualquier dirección.

**Características principales**

**•	Seis grados de libertad (6 DoF):** Permite los tres movimientos de traslación ((x, y, z) y los tres movimientos de rotación (roll, pitch, yaw).

**•	Sin eje ni límites definidos:** Como el movimiento es libre y omnidireccional, en el código XML de una junta floating no se configuran las etiquetas de eje (\<axis>) ni de límites (\<limit>).

**•	Conexión virtual:** No representa una pieza mecánica real (como un tornillo o una bisagra), sino una relación matemática de total libertad espacial.

  ![floating](imagenes_p1/floating.png)

#### Figura 11. Configuracion floating 


**¿Para qué sirve? (Aplicaciones comunes)**

Se utiliza principalmente para modelar la base de aquellos robots que no están fijos al suelo y pueden moverse libremente por el entorno de forma aérea, marítima o terrestre sin restricciones mecánicas:

**1. Drones y aeronaves (UAVs):** Permite que el cuerpo del dron se desplace y rote libremente en el espacio 3D respecto al origen del mundo.

**2. Robots humanoides y cuadrúpedos:** Se usa para conectar el torso o la pelvis del robot al "mundo", permitiendo simular cómo camina, salta o cae de forma libre.

**3. Vehículos submarinos (AUVs):** Modela el movimiento libre de navegación y buceo en entornos acuáticos.

**4. Sistemas espaciales / Satélites:** Para simular la flotación y movimiento en entornos de gravedad cero.

**Comportamiento en simulación**

Su comportamiento es muy particular, ya que gestiona la ubicación global del robot en el espacio:

**En RViz / ROS (Visualizador)**

**•	Publicación de Transformadas (TF):** RViz depende completamente de que un nodo externo (como un estimador de estado, odometría o un sistema de captura de movimiento) publique de forma constante la posición espacial y la orientación (usando cuaterniones) de la junta.

**•	Visualización libre:** Si los datos de transformadas son correctos, verás al robot "volar", desplazarse o rotar en la pantalla sin estar anclado a ningún punto fijo de la cuadrícula.

**•	Sin controles deslizantes:** Herramientas como joint_state_publisher_gui no pueden generar sliders para este tipo de junta, ya que no se puede controlar de forma lineal simple en la interfaz.

**En Gazebo (Simulador Físico)**

**•	Gravedad no restringida:** Al iniciar la simulación, si el robot está suspendido en el aire y tiene una junta floating respecto al mundo, el motor de física (ODE/Bullet) aplicará la gravedad de forma inmediata y el robot caerá libremente hasta chocar contra el suelo u otro objeto.

**•	Dinámica de cuerpo libre:** Responde con total realismo a cualquier fuerza externa. Si un objeto golpea al robot, este saldrá despedido, rotará o rebotará de acuerdo con su masa e inercias configuradas en \<inertial>.

**•	Actuación mediante fuerzas:** No se controla mediante motores de articulación tradicionales (ROS Control de juntas). En su lugar, para mover un robot con junta flotante en Gazebo, debes aplicar fuerzas y torques directamente al cuerpo rígido (como simular el empuje de las hélices de un dron o los propulsores de un submarino) a través de plugins específicos del simulador.


### Que es RVIZ

**RViz** (ROS Visualization) es una herramientas de visualización 3D de código abierto para el ecosistema Robot Operating System (ROS). Su función principal es mostrar en tiempo real lo que el robot "piensa" y percibe de su entorno a través de una interfaz gráfica interactiva.
A diferencia de un simulador de física, RViz traduce los datos numéricos de los algoritmos y sensores en elementos visuales comprensibles para los humanos. 

En términos sencillos: es la interfaz que traduce el código, los números y las matrices matemáticas del robot en una representación visual tridimensional (mapas, nubes de puntos, trayectorias) para que los ingenieros puedan entender exactamente qué está percibiendo y haciendo el sistema autónomo.

**Tres pilares de su definición técnica**

**•	Es un visualizador, no un simulador:** No calcula la gravedad, ni la fricción, ni simula un entorno físico (eso lo hace Gazebo). RViz solo dibuja de forma gráfica los mensajes y datos reales que el robot genera.

**•	Es una herramienta de telemetría:** Permite conectarse de forma remota a un robot físico a través de una red (como Wi-Fi) para observar su comportamiento de manera segura mientras opera en el mundo real.

**•	Es una interfaz de control básica:** Además de recibir información, permite interactuar de forma bidireccional con el robot, enviándole comandos gráficos directos como puntos de destino en un mapa o estimaciones de su posición inicial.

**Características Principales**

**•	Arquitectura Basada en Plugins:** Cada elemento visual (un mapa, una cámara, una trayectoria) es un módulo independiente. Puedes añadir, quitar o programar tus propios paneles de visualización.

**•	Visualización de Transformadas (TF):** Rastrea en tiempo real el árbol de coordenadas del robot. Muestra cómo se relacionan la base del robot (base_link), sus ruedas, cámaras y el mapa global (map).

**•	Herramientas de Interacción Directa:** Cuenta con herramientas integradas como 2D Nav Goal (para enviar coordenadas de destino al robot) y 2D Pose Estimate (para corregir manualmente la posición inicial del robot en un mapa).

**•	Renderizado de Múltiples Tipos de Datos:** Soporta de forma nativa nubes de puntos (PointCloud2), odometría, mallas poligonales, vectores de fuerza, imágenes de cámaras y marcadores personalizados creados por el usuario.

**Propiedades Técnicas (Tipos de Datos / Displays)**

En RViz, la información se organiza a través de propiedades llamadas Displays. Las más importantes son:

| Propiedad / Display | Descripción Técnica | Uso Típico |
| :--- | :--- | :--- |
| **RobotModel** | Carga el archivo URDF/XACRO del robot. | Ver el estado y movimiento de las articulaciones. |
| **LaserScan / PointCloud2** | Renderiza datos de distancia en 2D o nubes de puntos 3D. | Visualizar lo que detecta un LiDAR o una cámara Intel RealSense. |
| **Map / Costmap** | Muestra una rejilla de ocupación (celdas libres, ocupadas o desconocidas). | Navegación autónoma y evasión de obstáculos. |
| **Path / Odometry** | Dibuja líneas continuas o flechas en el espacio. | Visualizar la trayectoria planificada y el historial de movimiento. |
| **TF (Transform Synced)** | Muestra los ejes cartesianos (X-Rojo, Y-Verde, Z-Azul) de cada componente. | Verificar que los sensores estén calibrados en la posición correcta. |
| **Markers** | Permite al programador enviar formas geométricas básicas (esferas, cubos, texto). | Resaltar objetos detectados por algoritmos de Inteligencia Artificial. |

**Ventajas Competitivas**

**•	Depuración Eficiente (Debugging):** Reduce drásticamente el tiempo de desarrollo. En lugar de analizar millones de números de un sensor LiDAR en una terminal, ves la geometría del entorno en 3D al instante.

**•	Ligereza en Red:** RViz no procesa la física del mundo; solo dibuja datos. Esto permite ejecutar los algoritmos pesados en la computadora interna del robot y abrir RViz en una laptop externa conectada a la misma red Wi-Fi.

**•	Abstracción de Hardware:** Funciona exactamente igual si el robot es real o si los datos provienen de un simulador como Gazebo o Webots.

**•	Código Abierto y Comunidad:** Al ser el visualizador oficial de ROS/ROS 2, cuenta con miles de plugins creados por la comunidad y documentación masiva para resolver problemas


**Requisitos y Versiones**
**•	RViz vs RViz2:** La versión original funciona con ROS 1, mientras que RViz2 está rediseñada para trabajar de forma nativa en ROS 2.

**•	Entorno gráfico:** Al ser una interfaz pesada basada en pantallas, se debe ejecutar en la computadora de escritorio (como Ubuntu Desktop) y no dentro de la computadora interna del robot si esta carece de pantalla (Ubuntu Server).

**Diferencia clave: RViz vs. Gazebo**

Una confusión común en robótica es confundir RViz con Gazebo. Los desarrolladores originales lo resumen de la siguiente manera:

**•	Gazebo:** Te muestra lo que realmente ocurre en el mundo (simula la gravedad, las colisiones físicas y el entorno virtual).

**•	RViz:** Te muestra lo que el robot cree que está sucediendo según sus propios sensores y algoritmos. Si un sensor falla o un cálculo es erróneo, RViz mostrará ese error.

**Por qué es importante**

El **Unified Robot Description Format (URDF)** es el estándar de facto para describir modelos de robots en el ecosistema ROS y más allá. Define la estructura física de un robot — enlaces, articulaciones, geometría, propiedades inerciales y apariencia visual — en un único archivo XML que puede ser consumido por simuladores, planificadores de movimiento, herramientas de visualización y frameworks de control. Sin URDF, cada herramienta en el pipeline de robótica necesitaría su propio formato propietario de modelo de robot, haciendo la interoperabilidad casi imposible. URDF permite un flujo de trabajo de escribir una vez, usar en todas partes: el mismo archivo de modelo impulsa simulación en Gazebo, visualización en RViz, planificación de movimiento en MoveIt, y control de hardware real. Comprender URDF es esencial para cualquiera que construya, simule o despliegue robots en sistemas basados en ROS.

A continuación, se detallan los fragmentos de código e implementaciones más significativos:

**1. Modelado Geométrico y Visual de los Eslabones Añadidos** 

Con la finalidad de expandir la estructura original provista en la sesión, se programaron dos eslabones adicionales denominados link_2 (color verde) y link_3 (color azul). En el siguiente bloque de código se ilustra la implementación detallada de uno de los eslabones nuevos (link_2), garantizando la equivalencia entre las etiquetas visuales, de colisión y sus propiedades dinámicas estimadas:

<link name="link_2">
  <visual>
    <geometry>
      <box size="0.5 0.1 0.1"/>
    </geometry>
    <material name="green"/>
  </visual>
  <collision>
    <geometry>
      <box size="0.5 0.1 0.1"/>
    </geometry>
  </collision>
  <inertial>
    <mass value="1"/>
    <inertia ixx="1.0" ixy="0.0" ixz="0.0" iyy="1.0" iyz="0.0" izz="1.0"/>
  </inertial>
</link>

**Explicación técnica:** La etiqueta \<box> define una geometría prismática rectangular de (0.5 0.1 0.1) metros. Las propiedades de colisión clonan estas dimensiones exactas para prevenir fallos de penetración de mallas en el motor de simulación, mientras que el bloque \<inertial> provee un tensor de inercia genérico simétrico para habilitar cálculos dinámicos básicos.

**2. Articulación y Restricciones Cinemáticas Planares**

Para conectar jerárquicamente la cadena cinemática abierta, se integraron juntas de rotación que restringen el movimiento estrictamente a giros sobre el plano XY. Como muestra de cumplimiento, se presenta el código de la articulación link_2_joint:

<joint name="link_2_joint" type="revolute">
  <axis xyz="0 0 1"/>
  <limit effort="1000.0" lower="-3.14" upper="3.14" velocity="0.5"/>
  <origin rpy="0 0 0" xyz="0.45 0 0.1"/>
  <parent link="link_1"/>
  <child link="link_2"/>
</joint>

**Explicación técnica:** El atributo type="revolute" establece que la junta es angular y limitada. Al configurar \<axis xyz="0 0 1"/>, se obliga a que el giro ocurra exclusivamente en el eje coordenado Z, emulando los eslabones coplanares de un robot SCARA. Las etiquetas \<parent> y \<child> delimitan de manera estricta la relación padre-hijo dentro del árbol cinemático del sistema.

Mediante la modificación del archivo XML, se demostró la capacidad de expandir la descripción de un robot serial tipo SCARA añadiendo dos eslabones adicionales con juntas de revolución. Finalmente, la correcta integración del espacio de trabajo y el uso de archivos launch permitieron validar la estructura cinemática, las propiedades físicas y el diseño visual de forma interactiva en la plataforma RViz sin presentar errores.
